```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('houses.csv')
df.head()
```

```python
df.rename(columns = {'rate': 'price_per_sqft'},inplace = True)
```

```python
df.shape
```

```python
df.drop( columns = ['link' , 'property_id'], inplace = True)
```

```python
df.head()
```

```python
df.shape
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
df.info()
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

```python
#starting cleaning 
pd.set_option('display.max_rows', None)
df['society'].value_counts()
```

```python
df['society'].value_counts().shape
```

```python
# removing numbers and stars etc
import re
df['society'] = df['society'].apply(lambda name : re.sub(r'\d+(\.\d+)?\s?★','', str(name)).strip()).str.lower()
```

```python
df['society'].value_counts().shape
```

```python
df['society'] = df['society'].str.replace('nan','independent')
```

```python
df.head()
```

```python
# moving to price column
df['price'].value_counts()
```

```python
df = df[df['price'] != 'Price on Request']
```

```python
# function to treat prices in same unit
def treat_price(x):
    if type(x) == float:
        return x
    else:
        if x[1] == 'Lac':
            return round(float(x[0])/100,2)
        else:
            return round(float(x[0]),2)
```

```python
df['price'] = df['price'].str.split(' ').apply(treat_price)
```

```python
df.head()
```

```python
# moving to price_per_sqft
df['price_per_sqft'].value_counts()
```

```python
df['price_per_sqft'] = df['price_per_sqft'].str.split('/').str.get(0).str.replace('₹','').str.replace(',','').str.strip().astype('float')
```

```python
df.head()
```

```python
# moving to bedroom columns
df['bedRoom'].value_counts()
```

```python
df['bedRoom'] = df['bedRoom'].str.split(' ').str.get(0)
```

```python
df.head()
```

```python
# moving to bathroom column
df['bathroom'].value_counts()
```

```python
df['bathroom'] = df['bathroom'].str.split(' ').str.get(0)
```

```python
df.head()
```

```python
#moving to column balcony
df['balcony'].value_counts()
```

```python
df['balcony'] = df['balcony'].str.split(' ').str.get(0).replace('No','0')
```

```python
df.head()
```

```python
#moving to column additionalRoom
df['additionalRoom'].value_counts()
```

```python
df['additionalRoom'].fillna('not available',inplace=True)
```

```python
df['additionalRoom'].str.lower()
```

```python
df.head()
```

```python
# moving to column noOfFloor
df.rename( columns = {'noOfFloor':'floorNum'}, inplace = True)
```

```python
df['floorNum'].value_counts()
```

```python
df['floorNum'] = df['floorNum'].str.split(' ').str.get(0)
```

```python
df.head()
```

```python
#moving to column facing
df['facing'].value_counts()
```

```python
df['facing'].isnull().sum()
```

```python
df['facing'].fillna('NA',inplace = True)
```

```python
df.head()
```

```python
# changing area column by calculating on own
df['area'] = round((df['price']*10000000)/df['price_per_sqft'])
```

```python
# creating a new column of property_type
df.insert(loc=1,column='property_type',value='house')
```
