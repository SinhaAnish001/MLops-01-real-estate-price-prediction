```python
# removing numbers and stars etc
import re
df['society'] = df['society'].apply(lambda name : re.sub(r'\d+(\.\d+)?\s?★','', str(name)).strip()).str.lower()
```

```python
# moving to the next column price
df['price'].value_counts()
```

```python
df['price'].value_counts().shape
```

```python
df['price'].shape
```

```python
df= df[df['price'] != 'Price on Request']
```

```python
df['price'].shape
```

```python
# function to treaat prices in same unit i.e crore
def treat_price(x):
    if type(x) == float :
        return x
    else :
        if x[1] == 'Lac' :
            return round (float(x[0])/100,2)
        else :
            return round (float(x[0]),2)
```

```python
df['price'] = df['price'].str.split(' ').apply(treat_price)
```

```python
df.head()
```

```python
# moving to next column price_per_sqft
df['price_per_sqft'].value_counts()
```

```python
df['price_per_sqft'] = df['price_per_sqft'].str.split('/').str.get(0).str.replace('₹','').str.replace(',','').astype('float')
```

```python
df.head()
```

```python
df = df[~df['bedRoom'].isnull()]
```

```python
df['bedRoom'] = df['bedRoom'].str.split(' ').str.get(0).astype('int')
```

```python
# moving to next column bathroom
df['bathroom'].value_counts()
```

```python
df['bathroom'] = df['bathroom'].str.split(' ').str.get(0).astype('int')
```

```python
# moving to the next column balcony
df['balcony'].value_counts()
```

```python
df[df['balcony'].isnull()]
```

```python
df['balcony'] = df['balcony'].str.split(' ').str.get(0).str.replace('No','0')
```
